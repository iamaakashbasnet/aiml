# Machine Learning Pipeline
- Import the data
- Clean the data
- Split the data into traning set and test set
- Create a model (Choosing best fit algo)
- Train the model
- Make Predicition
- Evaluate the result and improve

```python
import pandas as pd
movies_data = pd.read_csv('movies.csv')
movies_data
```

```python
# Represents input data set (denoted by X)
X = movies_data.drop(columns=['genre'])
X
```

```python
# Represents output data set (denoted by y)
y = movies_data['genre']
y
```

```python
X.values
```

```python
y.values
```

```python
# Model creation
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()
model.fit(X.values, y.values)
```

```python
movies_data
```

```python
# Prediction
predictions = model.predict([[21, 1], [23, 0]])
predictions
```

```python
# Split training set and test set
from sklearn.model_selection import train_test_split

# More test data more accurate, changing test_size from 0.2 to 0.8 accuracy redued from 1.0 to 0.4
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
model.fit(x_train.values, y_train.values)
predictions = model.predict(x_test.values)
```

```python
# Accuracy eval (0-1)
from sklearn.metrics import accuracy_score

score = accuracy_score(y_test, predictions)
score
```

```python
# Model persistance
import joblib

joblib.dump(model, 'movie-recommender.joblib')
```

```python
# Loading the model
trained_model = joblib.load('movie-recommender.joblib')

new_predictions = trained_model.predict([[29,0]])
new_predictions
```

```python
# Visualizing how decision tree works
from sklearn import tree

tree.export_graphviz(trained_model, out_file='movies_recommender.dot',
                        feature_names=['age','gender'],
                        class_names=sorted(y.unique()),
                        label='all',
                        rounded=True,
                        filled=True
                    )
```
